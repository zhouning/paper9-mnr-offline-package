# Paper9 自然资源部成果可视化

检查预期输出、审计摘要、MPC 指标、变化统计，以及可在纯内网环境中使用的离线交互式 GIS 地图。

```python
import json
import os
from pathlib import Path

from IPython.display import HTML, display

from paper9_mnr.notebook_utils import project_root

ROOT = project_root()
CONFIG = os.environ.get("PAPER9_CONFIG", "configs/real_data_from_authority_slope.yml")
NOTEBOOK_OUTPUT_DIR = ROOT / "outputs/notebook"
NOTEBOOK_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"ROOT={ROOT}")
print(f"CONFIG={CONFIG}")
print(f"Notebook maps will be written to {NOTEBOOK_OUTPUT_DIR}")
```

## 输出清单与变化统计

```python
from paper9_mnr.notebook_utils import change_summary, output_inventory

inventory = output_inventory(CONFIG)
display(inventory)

changes = change_summary(CONFIG)
display(changes)
```

## 审计与 MPC 摘要

```python
from paper9_mnr.config import load_config

config = load_config(ROOT / CONFIG)
audit_path = ROOT / "outputs/audit_summary.json"
mpc_summary_path = ROOT / config["outputs"]["plan_dir"] / "mpc_summary.json"

if audit_path.exists():
    display(json.loads(audit_path.read_text(encoding="utf-8")))
else:
    print(f"Audit summary not found: {audit_path}")

if mpc_summary_path.exists():
    display(json.loads(mpc_summary_path.read_text(encoding="utf-8")))
else:
    print(f"MPC summary not found: {mpc_summary_path}")
```

## 交互式 GIS 地图

```python
from paper9_mnr.notebook_utils import input_layers_map_html, optimization_changes_map_html

input_map = NOTEBOOK_OUTPUT_DIR / "input_layers_map.html"
changes_map = NOTEBOOK_OUTPUT_DIR / "optimization_changes_map.html"

input_map_html = input_layers_map_html(CONFIG, max_features=1000, output_path=input_map)
display(HTML(input_map_html))
print(f"Saved: {input_map}")

try:
    changes_map_html = optimization_changes_map_html(CONFIG, max_features=1500, output_path=changes_map)
    display(HTML(changes_map_html))
    print(f"Saved: {changes_map}")
except FileNotFoundError as exc:
    print(exc)
```
